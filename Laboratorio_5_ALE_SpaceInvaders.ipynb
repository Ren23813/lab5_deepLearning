## 5. Agente aleatorio en `ALE/SpaceInvaders-v5` (con video)

Generar y guardar un video de 1 episodio completo del agente aleatorio
jugando Space Invaders. También se muestran: número de pasos sobrevividos y la
recompensa total obtenida.

```python
resultado_random = generar_video_agente(
    nombre_entorno="ALE/SpaceInvaders-v5",
    funcion_agente=agente_aleatorio,
    video_folder="videos/agente_aleatorio",
    name_prefix="space_invaders_random",
    n_episodios=1,
)

print("Videos generados:")
for v in resultado_random["videos"]:
    print(" -", v)

print()
print("Métricas por episodio (agente aleatorio):")
for i, m in enumerate(resultado_random["metricas"]):
    print(f"Episodio {i}: pasos={m['pasos']}, recompensa_total={m['recompensa_total']}, "
          f"terminated={m['terminated']}, truncated={m['truncated']}")
```

### Reproducir el video del agente aleatorio en el notebook

```python
from IPython.display import Video

Video(resultado_random["videos"][0], embed=True, width=480)
```
